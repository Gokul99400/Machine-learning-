# 🚀 Mini Project: Student Performance Analyzer

Use the following dataset.

```python
project_data = {
    "Name": ["Arun", "Priya", "Rahul", "Divya", "Kiran", "Meena", "Vijay", "Anu"],
    "Department": ["AI", "DS", "AI", "CSE", "DS", "AI", "CSE", "DS"],
    "Marks": [85, 92, 67, 78, 95, 72, 88, 81],
    "Attendance": [90, 95, 75, 82, 98, 79, 91, 85]
}

students = pd.DataFrame(project_data)
students
```

### Task 1: Display the first 5 students.

```python
display(students.head(5))
```

### Task 2: Display only `Name` and `Marks`.

```python
display(students[['Name', 'Marks']])
```

### Task 3: Find students who scored more than 80.

```python
display(students[students['Marks'] > 80])
```

### Task 4: Find students with attendance greater than 85.

```python
display(students[students['Attendance'] > 85])
```

### Task 5: Find students with marks > 80 AND attendance > 85.

```python
display(students[(students['Marks'] > 80) & (students['Attendance'] > 85)])
```

### Task 6: Find the average marks.

```python
print(f"Average Marks: {students['Marks'].mean()}")
```

### Task 7: Find the highest-scoring student.

```python
display(students.loc[students['Marks'].idxmax()])
```

### Task 8: Find average marks for each department.

### Task 9: Add a `Performance` column: `Excellent` if marks >= 90, otherwise `Good`.

```python
import numpy as np
students['Performance'] = np.where(students['Marks'] >= 90, 'Excellent', 'Good')
display(students)
```
